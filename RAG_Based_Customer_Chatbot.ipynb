```python
from langchain_community.document_loaders import PyPDFLoader  # this is to load the data
from langchain_text_splitters import RecursiveCharacterTextSplitter # this is to split the data --. tokens. 
from langchain_community.embeddings import HuggingFaceEmbeddings # this is to convert the data in to Embeddings
from langchain_community.vectorstores import FAISS  # We need a vecotr data base to store our source FAISS is open source vector database (META)
from langchain_ollama import OllamaLLM # This is the LLM which we are going to use. Open Source
import warnings
warnings.filterwarnings('ignore')
# imported all the libraries that are needed
```

```python
loader = PyPDFLoader("customer_support.pdf")
docs = loader.load()
# This is the Knowledge Source we are loadign it
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_documents(docs)
# we are dividing the huge data on to small chunks / pieces
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
# we are converting the data from text to numerical format called as embedding / vectors
```

```python
vectorstore = FAISS.from_documents(chunks, embeddings)

# We are storing these knowledge source data in to some vector database from where we can search and retrieve the relevant chunks
```

```python
retriever = vectorstore.as_retriever()
```

```python
llm = OllamaLLM(model="llama3")  # llm is like a person who can read,understand and write
```

```python
query = str(input())  # user will ask the question
docs = retriever.invoke(query)   # retriver will convert the question in to embedding and go and search for the relevant chunks in vectorDB 
context = "\n".join([d.page_content for d in docs]) # this is the retrieve information
prompt = f"""
You are a customer support assistant. Answer the customer's question using only the context below.
If the answer is not present in the context, say you could not find that information and suggest
contacting human support.

Context:
{context}

Question:
{query}

Answer:
"""
response = llm.invoke(prompt)
print('Chatbot Response:',response)
```
